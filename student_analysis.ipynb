```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# Student Performance Analysis
This notebook analyzes student performance data using Python, Pandas, Matplotlib, and Seaborn.


```python
df = pd.read_csv('/content/student-mat.csv', sep=';')
```

```python
df.head()
```

## Dataset Exploration
Checking dataset dimensions, columns, and data types.

```python
print("Shape of dataset:")
print(df.shape)
```

```python
print("Columns:")
print(df.columns)
```

```python
print("Data Types:")
print(df.dtypes)
```

## Missing Values
Checking if any values are missing.

```python
df.isnull().sum()
```

## Duplicate Records
Checking duplicate rows.

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

## Average Final Grade
Calculating average final grade of students.

```python
avg_grade = df['G3'].mean()
print("Average Final Grade:", avg_grade)
```

## Students Scoring Above 15
Finding students with excellent performance.

```python
above_15 = df[df['G3'] > 15]

print("Number of Students Scoring Above 15:")
print(len(above_15))
```

## Correlation Between Study Time and Performance
Checking relationship between study hours and grades.

```python
correlation = df['studytime'].corr(df['G3'])

print("Correlation between Study Time and G3:")
print(correlation)
```

A positive value indicates that students who study more tend to score higher.

## Average Grade by Gender
Comparing male and female student performance.

```python
gender_avg = df.groupby('sex')['G3'].mean()

print(gender_avg)
```

## Histogram of Final Grades
Visualizing grade distribution.

```python
plt.figure(figsize=(8,5))

plt.hist(df['G3'], bins=10)

plt.title('Distribution of Final Grades')
plt.xlabel('Grade')
plt.ylabel('Frequency')

plt.show()
```

## Study Time vs Grades
Scatter plot showing relationship between study time and final grade.

```python
plt.figure(figsize=(8,5))

plt.scatter(df['studytime'], df['G3'])

plt.title('Study Time vs Final Grade')
plt.xlabel('Study Time')
plt.ylabel('Final Grade')

plt.show()
```

## Male vs Female Average Score

```python
gender_avg.plot(kind='bar')

plt.title('Average Final Grade by Gender')
plt.xlabel('Gender')
plt.ylabel('Average Grade')

plt.show()
```
